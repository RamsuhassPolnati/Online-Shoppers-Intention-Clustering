Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaled_data = scaler.fit_transform(data)
```

```python
scaled_data = pd.DataFrame(scaled_data, columns = data.columns)
```

```python
scaled_data.head()
```

Determining the Optimal Number of Clusters

```python
# Apply the Elbow Method to identify the optimal number of clusters.

from sklearn.cluster import KMeans

wcss = []
k_range = range(1, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)

    kmeans.fit(scaled_data)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(k_range, wcss, marker="o")

plt.title("Elbow Method for Optimal K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Within Cluster Sum of Squares (WCSS)")
plt.xticks(k_range)
plt.grid(True)

plt.savefig(r"C:\Users\ramsu\Desktop\OnlineShoppersIntention\images\elbow_method.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Calculate Silhouette Scores
silhouette_scores = []
k_range = range(2, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)

    labels = kmeans.fit_predict(scaled_data)

    score = silhouette_score(scaled_data, labels)
    silhouette_scores.append(score)

    print(f"K = {k}, Silhouette Score = {score:.4f}")

# Plot Silhouette Scores
plt.figure(figsize=(8, 5))

plt.plot(k_range, silhouette_scores, marker="o")

plt.title("Silhouette Score for Optimal K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.xticks(k_range)
plt.grid(True)

plt.savefig(r"C:\Users\ramsu\Desktop\OnlineShoppersIntention\images\silhouette_score.png", dpi=300, bbox_inches="tight")

plt.show()
```

K-Means Clustering

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters = 2 , random_state = 15,n_init=10)
```

```python
clusters = kmeans.fit(scaled_data)
clusters
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(scaled_data.iloc[:, 0], scaled_data.iloc[:, 1], c=clusters.labels_, cmap="viridis")
plt.show()
```

```python
from sklearn.decomposition import PCA
dec_model = PCA(n_components = 2)
x_pca = dec_model.fit_transform(scaled_data)
```

Cluster Visualization

```python
plt.figure(figsize=(8,6))
plt.scatter(x_pca[:,0], x_pca[:,1], c=clusters.labels_, cmap="viridis")
plt.title("K-Means Clusters Using PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.savefig(r"C:\Users\ramsu\Desktop\OnlineShoppersIntention\images\kmeans_pca_clusters.png", dpi=300, bbox_inches="tight")
plt.show()
```
